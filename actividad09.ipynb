GRAFICOS

hacemos un grafico pie en base a los valores numericos de tap

```python
h = areas["tap"].value_counts()
px.pie(data_frame = h, values = h.values, names = h.index, title = "pie de tap").show()
```

hacemos barra obtiene los primeros valores de "sag"

```python
h = areas["sag"].head(5).value_counts()
px.bar(data_frame = h, x = h.values, y = h.index, title = "barra de sag").show()
```
