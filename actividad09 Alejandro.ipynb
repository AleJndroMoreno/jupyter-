```python
import folium as mapp
```

```python
g = areas[areas["tap"] == 2]

mapa = mapp.Map(
    location=[g["lat"].mean(), g["lng"].mean()],
    zoom_start=4
)
colors = {
    1: "purple",
    2: "red",
    3: "green",
    4: "blue"
}
for _, fila in g.iterrows():

    mapp.CircleMarker(
        location=[fila["lat"], fila["lng"]],
        radius=6,
        color=colors.get(fila["jap"], "gray"),
        fill=True,
        fill_color=colors.get(fila["jap"], "gray"),
        popup=fila["nam"]
    ).add_to(mapa)
mapa
```
